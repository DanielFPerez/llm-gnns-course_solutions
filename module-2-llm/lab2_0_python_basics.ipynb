## 2 · Variables and basic types

A **variable** is a name that points to a value. Python figures out the type automatically — you never have to declare it.

| Type | Example | Meaning |
|---|---|---|
| `int` | `42` | whole number |
| `float` | `3.14` | decimal number |
| `str` | `"hello"` | text (single or double quotes both work) |
| `bool` | `True` / `False` | logical value |

Use `type(x)` to check what type a variable holds.

````python
customer_name = "Ana"          # str
monthly_charges = 79.50        # float
tenure_months = 14             # int
is_active = True               # bool

print(customer_name, type(customer_name))
print(monthly_charges, type(monthly_charges))
print(tenure_months, type(tenure_months))
print(is_active, type(is_active))
````

Basic arithmetic works as you'd expect: `+ - * /` for the usual operations, `**` for power, `//` for integer (floor) division, and `%` for remainder.

````python
total_paid = monthly_charges * tenure_months
print(f"Total paid so far: {total_paid}")

print(7 / 2)    # true division -> 3.5
print(7 // 2)   # floor division -> 3
print(7 % 2)    # remainder -> 1
print(2 ** 10)  # power -> 1024
````

## 4 · Control flow: `if` / `elif` / `else`

Python decides which block of code to run based on conditions. **Indentation (spaces) is not optional in Python — it defines which lines belong to which block.** Most editors indent automatically with 4 spaces after a `:`.

Comparison operators: `==` (equal), `!=` (not equal), `<`, `>`, `<=`, `>=`. Combine conditions with `and`, `or`, `not`.

````python
monthly_charges = 85.0

if monthly_charges > 80:
    risk = "high"
elif monthly_charges > 50:
    risk = "medium"
else:
    risk = "low"

print(f"Billing risk category: {risk}")
````

## 5 · Lists

A **list** is an ordered, changeable collection, written with square brackets. Items are accessed by **index**, starting at `0` (this trips up almost everyone at first — the first element is `my_list[0]`, not `my_list[1]`).

````python
tenures = [14, 3, 27, 1, 45]

print(tenures[0])       # first element
print(tenures[-1])      # last element
print(tenures[1:3])     # slice: index 1 up to (not including) 3
print(len(tenures))     # number of elements

tenures.append(8)       # add an element at the end
print(tenures)

for t in tenures:
    print(f"Customer tenure: {t} months")
````

### Exercise 2.0.4 `[Core]` — Loop and accumulate

Given the list `charges` below, use a `for` loop to compute the **total** and the **average** monthly charge manually (i.e. without using the built-in `sum()` — though feel free to double-check your answer with it afterwards).

````python
# --- SOLUTION ---
charges = [79.5, 45.0, 103.2, 60.0, 88.75]

total = 0
for c in charges:
    total += c   # shorthand for: total = total + c

average = total / len(charges)

print(f"Total: {total:.2f}")
print(f"Average: {average:.2f}")
print(f"Check with built-in sum(): {sum(charges):.2f}")
````

## 6 · Dictionaries

A **dictionary** stores `key: value` pairs, written with curly braces. Instead of accessing items by numeric position (like a list), you access them by **key** — this is how you'll usually represent a single record, like one customer or one row of a table.

````python
customer = {
    "name": "Ana",
    "tenure_months": 14,
    "monthly_charges": 79.5,
    "churned": False,
}

print(customer["name"])
print(customer["monthly_charges"])

customer["monthly_charges"] = 82.0   # update a value
customer["contract"] = "month-to-month"  # add a new key

for key, value in customer.items():
    print(f"{key}: {value}")
````

## 7 · Putting it together: lists of dictionaries, and functions

Real datasets are usually a **list of records**, where each record is a dictionary — exactly the shape you'll get back from many APIs and, conceptually, what a DataFrame row looks like. A **function** lets you package logic under a name so you can reuse it instead of repeating code.

```python
def function_name(parameter1, parameter2):
    """Optional docstring describing what the function does."""
    result = ...        # do something
    return result
```

````python
customers = [
    {"name": "Ana", "tenure_months": 14, "monthly_charges": 79.5, "churned": False},
    {"name": "Luis", "tenure_months": 2, "monthly_charges": 95.3, "churned": True},
    {"name": "Marta", "tenure_months": 40, "monthly_charges": 55.0, "churned": False},
]

def average_charge(records):
    """Return the average monthly_charges across a list of customer dicts."""
    total = sum(r["monthly_charges"] for r in records)
    return total / len(records)

print(f"Average monthly charge: {average_charge(customers):.2f}")
````

### Exercise 2.0.6 `[Core]` — Write your own function

Write a function `high_value_customers(records, threshold=70)` that returns a **list of names** of customers whose `monthly_charges` is above `threshold`. Test it on the `customers` list above.

````python
# --- SOLUTION ---
def high_value_customers(records, threshold=70):
    """Return names of customers with monthly_charges above threshold."""
    names = []
    for r in records:
        if r["monthly_charges"] > threshold:
            names.append(r["name"])
    return names

print(high_value_customers(customers))
print(high_value_customers(customers, threshold=90))
````

### Exercise 2.0.7 `[Extension]` — List comprehensions

A **list comprehension** is a compact way to build a list from a loop + optional condition, in one line: `[expression for item in iterable if condition]`. Rewrite `high_value_customers` from the previous exercise as a one-line list comprehension.

````python
# --- SOLUTION ---
def high_value_customers_v2(records, threshold=70):
    return [r["name"] for r in records if r["monthly_charges"] > threshold]

print(high_value_customers_v2(customers))
assert high_value_customers_v2(customers) == high_value_customers(customers)
print("Matches the loop-based version.")
````

## 8 · `[Extension]` A peek at external libraries

Everything above is plain Python — no installs needed. The rest of the course relies heavily on **libraries**: pre-written code you `import` and reuse. `pandas` (Module I) and `numpy` are two you've already seen; here's the connection to what you just learned.

A NumPy **array** looks similar to a list, but supports fast, vectorized math — operations apply to every element at once, without writing a `for` loop yourself.

````python
import numpy as np

charges_list = [79.5, 45.0, 103.2, 60.0, 88.75]
charges_arr = np.array(charges_list)

print(charges_arr)
print(charges_arr.mean())        # same idea as our manual average, no loop needed
print(charges_arr * 1.1)         # apply a 10% increase to every element at once
print(charges_arr[charges_arr > 70])   # keep only elements above 70 — like our list comprehension
````

> **Notice the pattern:** `charges_arr[charges_arr > 70]` does conceptually the same thing as the list comprehension in Exercise 2.0.7 — filter by a condition — but NumPy (and later, pandas) let you express it without an explicit loop. You'll see this idiom constantly from here on.

## 9 · Common beginner pitfalls (reference)

Keep this list handy — almost every early bug traces back to one of these:

| Symptom | Likely cause |
|---|---|
| `IndentationError` | Mismatched spaces/tabs, or forgot the `:` before an indented block |
| `list index out of range` | Forgot indexing starts at `0`, or looped one step too far |
| A cell "doesn't see" a variable | You never ran the cell that defines it — run cells top to bottom |
| `NameError: name 'x' is not defined` | Typo in the variable name, or used it before creating it |
| `TypeError: can only concatenate str (not "int") to str` | Mixed a string and a number, e.g. `"Age: " + 30` — use an f-string instead |
| A function "did nothing" | Forgot the `return` statement — the function ran but gave the value back to no one |
| `=` vs `==` | `=` **assigns** a value; `==` **compares** two values. `if x = 5:` is a syntax error — you want `if x == 5:` |
